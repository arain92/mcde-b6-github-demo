# Q35. Student Marks
Question: Find the highest, lowest, and average marks and the top-scoring student.

```python
marks = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zain": 88
}

print("Highest:", max(marks.values()))
print("Lowest:", min(marks.values()))
print("Average:", sum(marks.values()) / len(marks))
print("Top Student:", max(marks, key=marks.get))
```

# Q36. Nested Employee Records
Question: Extract Sara's salary, increase Ali's salary by 15%, and calculate the total payroll.

```python
employees = {
    "E101": {"name": "Ali", "salary": 80000},
    "E102": {"name": "Sara", "salary": 120000},
    "E103": {"name": "Ahmed", "salary": 95000}
}

print("Sara's Salary:", employees["E102"]["salary"])

employees["E101"]["salary"] *= 1.15

total = (
    employees["E101"]["salary"]
    + employees["E102"]["salary"]
    + employees["E103"]["salary"]
)

print("Ali's New Salary:", employees["E101"]["salary"])
print("Total Payroll:", total)
```

#  Q38. Order Calculation
Question: Calculate each item's subtotal, grand total, and total after a 10% discount.

```python
order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}

laptop = order["laptop"]["price"] * order["laptop"]["quantity"]
mouse = order["mouse"]["price"] * order["mouse"]["quantity"]
keyboard = order["keyboard"]["price"] * order["keyboard"]["quantity"]

total = laptop + mouse + keyboard
discount = total * 0.10
final_total = total - discount

print("Laptop:", laptop)
print("Mouse:", mouse)
print("Keyboard:", keyboard)
print("Grand Total:", total)
print("Discount:", discount)
print("Final Total:", final_total)
```

# Q39. Employee Performance
Question: Calculate each employee's average, find the top performer, and find the highest SQL score.

```python
employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

ali_avg = (90 + 85 + 88) / 3
sara_avg = (95 + 92 + 90) / 3
ahmed_avg = (80 + 87 + 84) / 3

print("Ali Average:", ali_avg)
print("Sara Average:", sara_avg)
print("Ahmed Average:", ahmed_avg)

top = max(
    {"Ali": ali_avg, "Sara": sara_avg, "Ahmed": ahmed_avg},
    key={"Ali": ali_avg, "Sara": sara_avg, "Ahmed": ahmed_avg}.get
)

print("Top Performer:", top)
print("Highest SQL Score:", max(
    employees["Ali"]["sql"],
    employees["Sara"]["sql"],
    employees["Ahmed"]["sql"]
))
```

# Q40. Inventory Management
Question: Calculate total inventory value, find the most valuable item by price × stock, and update stock after selling two laptops and five mice.

```python
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

laptop_value = 120000 * 5
mouse_value = 1500 * 20
keyboard_value = 3500 * 10

total_value = laptop_value + mouse_value + keyboard_value

print("Total Inventory Value:", total_value)
print("Most Valuable Item: Laptop")

inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5

print("Updated Laptop Stock:", inventory["P101"]["stock"])
print("Updated Mouse Stock:", inventory["P102"]["stock"])
print("Updated Inventory:", inventory)
```
