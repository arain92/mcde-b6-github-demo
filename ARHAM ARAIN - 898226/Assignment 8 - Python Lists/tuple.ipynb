# Q11. Tuple Indexing
Extract the second, last, and middle three elements from the tuple.

```python

colors = ("red", "green", "blue", "yellow", "purple")

second = colors[1]
last = colors[-1]
middle_three = colors[1:4]

print(second)
print(last)
print(middle_three)
```

# Q13. Tuple Concatenation
 Combine two tuples and create another tuple in reverse order.

```python

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

combined = tuple1 + tuple2
reversed_tuple = combined[::-1]

print("Combined:", combined)
print("Reversed:", reversed_tuple)
```

# Q15. Extended Tuple Unpacking
 Unpack the first two and last elements into variables, and store the remaining elements in a list.

```python

data = (100, 200, 300, 400, 500, 600)

first, second, *middle, last = data

print("First:", first)
print("Second:", second)
print("Middle:", middle)
print("Last:", last)
```

# Q16. Modify an Immutable Tuple
 Create a new tuple by replacing 30 with 300.

```python

numbers = (10, 20, 30, 40, 50)

new_numbers = numbers[:2] + (300,) + numbers[3:]

print("Original:", numbers)
print("New tuple:", new_numbers)
```

# Q17. Nested Tuple Access
Extract Sara's second score, Ali's average score, and Ahmed's highest score.

```python

students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

sara_second = students[1][1][1]

ali_scores = students[0][1]
ali_average = sum(ali_scores) / len(ali_scores)

ahmed_scores = students[2][1]
ahmed_highest = max(ahmed_scores)

print("Sara's second score:", sara_second)
print("Ali's average:", ali_average)
print("Ahmed's highest score:", ahmed_highest)
```

# Q18. Tuple Sorting
 Sort employees by salary in descending order without loops, comprehensions, or lambda.

```python

from operator import itemgetter

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sorted_employees = sorted(
    employees,
    key=itemgetter(1),
    reverse=True
)

print(sorted_employees)
```

# Q19. Tuple Matrix
 Extract the diagonal, calculate its product, and create a tuple of row sums.

```python

matrix = (
    (2, 4, 6),
    (8, 10, 12),
    (14, 16, 18)
)

diagonal = (
    matrix[0][0],
    matrix[1][1],
    matrix[2][2]
)

diagonal_product = (
    diagonal[0] * diagonal[1] * diagonal[2]
)

row_sums = (
    sum(matrix[0]),
    sum(matrix[1]),
    sum(matrix[2])
)

print("Diagonal:", diagonal)
print("Diagonal product:", diagonal_product)
print("Row sums:", row_sums)
```

# Q20. Inventory Tuple Challenge
 Calculate the total inventory value, find the most expensive product, and create a reversed tuple of product names.

```python

from operator import itemgetter

records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

total_value = (
    records[0][1] * records[0][2]
    + records[1][1] * records[1][2]
    + records[2][1] * records[2][2]
)

most_expensive = max(records, key=itemgetter(1))

names = (
    records[0][0],
    records[1][0],
    records[2][0]
)

reversed_names = names[::-1]

print("Total inventory value:", total_value)
print("Most expensive product:", most_expensive[0])
print("Reversed names:", reversed_names)
```
