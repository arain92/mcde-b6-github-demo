# Q43. Sales Data
Question: Calculate revenue, top-revenue product, a name-to-revenue dictionary, and a set of prices.

```python
sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

laptop = sales[0][1] * sales[0][2]
mouse = sales[1][1] * sales[1][2]
keyboard = sales[2][1] * sales[2][2]

revenue = laptop + mouse + keyboard

revenues = {
    sales[0][0]: laptop,
    sales[1][0]: mouse,
    sales[2][0]: keyboard
}

prices = {sales[0][1], sales[1][1], sales[2][1]}

print("Total Revenue:", revenue)
print("Top Product:", max(revenues, key=revenues.get))
print("Revenue Dictionary:", revenues)
print("Prices:", prices)
```

# Q45. Course Enrollment
Question: Find students enrolled in exactly two courses, all three courses, exactly one course, and count unique students.

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

python = courses["Python"]
sql = courses["SQL"]
aws = courses["AWS"]

exactly_two = (
    (python & sql) - aws
) | (
    (python & aws) - sql
) | (
    (sql & aws) - python
)

all_three = python & sql & aws

all_students = python | sql | aws

exactly_one = all_students - (python & sql) - (python & aws) - (sql & aws)

print("Exactly Two:", exactly_two)
print("All Three:", all_three)
print("Exactly One:", exactly_one)
print("Unique Students:", len(all_students))
```

# Q49. Product Recommendations
Question: Find products purchased by at least two users, Ali's wishlist excluding purchases, shared wishlist products, and all products.

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"},
            "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"},
             "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"},
              "wishlist": {"Monitor", "Headphones"}}
}

bought_twice = (
    users["Ali"]["purchased"] & users["Ahmed"]["purchased"]
) | (
    users["Ali"]["purchased"] & users["Sara"]["purchased"]
) | (
    users["Sara"]["purchased"] & users["Ahmed"]["purchased"]
)

ali_wishlist = (
    users["Ali"]["wishlist"] - users["Ali"]["purchased"]
)

shared_wishlist = (
    (users["Ali"]["wishlist"] & users["Sara"]["wishlist"])
    | (users["Ali"]["wishlist"] & users["Ahmed"]["wishlist"])
    | (users["Sara"]["wishlist"] & users["Ahmed"]["wishlist"])
)

all_products = (
    users["Ali"]["purchased"] | users["Sara"]["purchased"]
    | users["Ahmed"]["purchased"] | users["Ali"]["wishlist"]
    | users["Sara"]["wishlist"] | users["Ahmed"]["wishlist"]
)

print("Bought by at least two:", bought_twice)
print("Ali's Recommendations:", ali_wishlist)
print("Shared Wishlist:", shared_wishlist)
print("All Products:", all_products)
```

# Q50. Final Company Analysis
Question: Calculate payroll per department, company payroll, highest-paid employee, shared skills, Engineering-only skills, remaining budgets, higher average salary, name-to-salary dictionary, all unique skills, and whether the combined budget is sufficient.

```python
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

eng = company["departments"]["Engineering"]
ana = company["departments"]["Analytics"]

eng_payroll = eng["employees"][0][1] + eng["employees"][1][1]
ana_payroll = ana["employees"][0][1] + ana["employees"][1][1]
total_payroll = eng_payroll + ana_payroll

print("Engineering Payroll:", eng_payroll)
print("Analytics Payroll:", ana_payroll)
print("Company Payroll:", total_payroll)
print("Highest Paid:", "Sara", 150000)

shared = (
    eng["employees"][0][2] & eng["employees"][1][2]
    & ana["employees"][0][2] & ana["employees"][1][2]
)

eng_skills = eng["employees"][0][2] | eng["employees"][1][2]
ana_skills = ana["employees"][0][2] | ana["employees"][1][2]

print("Shared Skills:", shared)
print("Engineering-only:", eng_skills - ana_skills)

print("Engineering Remaining:", eng["budget"] - eng_payroll)
print("Analytics Remaining:", ana["budget"] - ana_payroll)

print("Higher Average Salary:", "Engineering")

salaries = {
    "Ali": 120000,
    "Sara": 150000,
    "Ahmed": 100000,
    "Zain": 130000
}
print("Name-to-Salary:", salaries)

all_skills = eng_skills | ana_skills
print("All Skills:", all_skills)

budget = eng["budget"] + ana["budget"]
print("Budget Sufficient:", budget >= total_payroll)
```
