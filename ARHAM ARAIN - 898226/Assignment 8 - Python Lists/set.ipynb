# Q21. Unique Values
 Convert the list into a set and count unique elements.

```python

numbers = [10, 20, 10, 30, 40, 20, 50]

unique_numbers = set(numbers)
count = len(unique_numbers)

print("Unique values:", unique_numbers)
print("Count:", count)
```

# Q22. Set Operations
 Find the union, intersection, difference, and symmetric difference of two sets.

```python

a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

union = a | b
intersection = a & b
difference = a - b
symmetric_difference = a ^ b

print("Union:", union)
print("Intersection:", intersection)
print("A - B:", difference)
print("Symmetric difference:", symmetric_difference)
```

# Part 2: Intermediate Level
# Q24. Common Students
 Find students taking both courses and students taking exactly one course.

```python

python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hamza", "Usman"}

both_courses = python_students & sql_students
one_course = python_students ^ sql_students

print("Both courses:", both_courses)
print("Exactly one course:", one_course)
```

# Q25. Subset and Superset
 Check whether required skills are a subset and candidate skills are a superset.

```python

required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}

is_subset = required_skills.issubset(candidate_skills)
is_superset = candidate_skills.issuperset(required_skills)

print("Required skills are subset:", is_subset)
print("Candidate skills are superset:", is_superset)
```

# Q27. Three-Set Analysis
 Find students in all three sets, exactly in Python and SQL but not AWS, and in at least two sets.

```python

python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

all_three = python & sql & aws

python_sql_only = (python & sql) - aws

at_least_two = (
    (python & sql)
    | (python & aws)
    | (sql & aws)
)

print("All three:", all_three)
print("Python and SQL only:", python_sql_only)
print("At least two:", at_least_two)
```

# Q28. Transaction Analysis
Short English question: Find transaction IDs appearing on all days, exactly one day, and at least two days

```python

day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

all_days = day1 & day2 & day3

exactly_one_day = (
    (day1 - day2 - day3)
    | (day2 - day1 - day3)
    | (day3 - day1 - day2)
)

at_least_two_days = (
    (day1 & day2)
    | (day1 & day3)
    | (day2 & day3)
)

print("All days:", all_days)
print("Exactly one day:", exactly_one_day)
print("At least two days:", at_least_two_days)
```

# Q29. Access Control
 Find missing and extra permissions and check full authorization.

```python

required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing = required_permissions - user_permissions
extra = user_permissions - required_permissions

authorized = required_permissions.issubset(user_permissions)

print("Missing permissions:", missing)
print("Extra permissions:", extra)
print("Fully authorized:", authorized)
```

# Q30. Customer Segmentation
 Find customers using exactly one, exactly two, and all three platforms.

```python

website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website & mobile & store

exactly_two = (
    ((website & mobile) - store)
    | ((website & store) - mobile)
    | ((mobile & store) - website)
)

exactly_one = (
    (website - mobile - store)
    | (mobile - website - store)
    | (store - website - mobile)
)

print("Exactly one platform:", exactly_one)
print("Exactly two platforms:", exactly_two)
print("All three platforms:", all_three)
```
